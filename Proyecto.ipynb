# Descarga del Dataset UrbanSound8K

```python
import soundata

# esto lo hice porque en mi ssd ya no hay espacio xd
ruta_descarga = r"D:\Usuarios\teso\Documentos\IPN\SextoSemestre\Computo Paralelo\Proyecto_Paralelo\dataset"

dataset = soundata.initialize('urbansound8k', data_home=ruta_descarga)

#dataset.download()  # Descargará el dataset en la ruta_descarga
dataset.validate()  # Validará los archivos en esa misma ruta

example_clip = dataset.choice_clip()  # Elige un clip de ejemplo al azar
print(example_clip)  # Muestra la información del clip
```

# Exploración Básica

```python
# 1. Obtener la lista de IDs y seleccionar el primero
ids_clips = dataset.clip_ids
primer_clip_id = ids_clips[0]
clip = dataset.clip(primer_clip_id)

print("\n--- Metadatos del Clip ---")
print(f"ID del clip: {clip.clip_id}")

# clip.tags.labels devuelve una lista, y tomamos el primer (y único) elemento [0]
etiqueta_clase = clip.tags.labels[0]
print(f"Clase a la que pertenece: {etiqueta_clase}") 

# El número de 'Fold' usualmente determina a qué subcarpeta de prueba pertenece (1-10)
print(f"Carpeta de validación (Fold): {clip.fold}") 

# 2. Inspeccionar las propiedades del archivo de audio real
audio_array, sample_rate = clip.audio

print("\n--- Propiedades del Audio ---")
print(f"Frecuencia de muestreo (Sample Rate): {sample_rate} Hz")
print(f"Forma del arreglo numérico (Samples, Canales): {audio_array.shape}")
```

# Análisis global de clases

```python
import pandas as pd
import os

ruta_descarga = r"D:\Usuarios\teso\Documentos\IPN\SextoSemestre\Computo Paralelo\Proyecto_Paralelo\dataset"
ruta_metadata = None

# 1. Buscar automáticamente el archivo UrbanSound8K.csv en todas las subcarpetas
for directorio_raiz, subcarpetas, archivos in os.walk(ruta_descarga):
    for archivo in archivos:
        if archivo == 'UrbanSound8K.csv' or archivo.endswith('.csv'):
            ruta_metadata = os.path.join(directorio_raiz, archivo)
            break
    if ruta_metadata:
        break

# 2. Verificar si lo encontró y leerlo
if ruta_metadata:
    print(f"¡Archivo encontrado en: {ruta_metadata}!\n")
    
    # Cargar los datos
    df = pd.read_csv(ruta_metadata)

    print("--- Resumen General del Dataset ---")
    print(df.head())

    print("\n--- Cantidad de audios por Clase ---")
    print(df['class'].value_counts())

    print("\n--- Distribución por Fold (Carpetas) ---")
    print(df['fold'].value_counts())
else:
    print(f"No se encontró el archivo CSV dentro de: {ruta_descarga}")
    print("Revisa manualmente la carpeta para confirmar que la descarga se completó correctamente.")
```
