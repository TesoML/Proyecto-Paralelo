```python
import os
import time
import numpy as np
import pandas as pd
import librosa
from tqdm.auto import tqdm

def extraer_stft_y_guardar_csv(ruta_base, df_metadata, ruta_salida_csv, tamaño_lote=500):
    "Extrae características de la STFT de forma secuencial."
    
    tiempo_inicio = time.perf_counter()
    archivo_csv_existe = False
    
    total_archivos = len(df_metadata)
    
    # Procesar por lotes para proteger la memoria RAM
    for inicio in range(0, total_archivos, tamaño_lote):
        fin = min(inicio + tamaño_lote, total_archivos)
        lote_df = df_metadata.iloc[inicio:fin]
        
        print(f"\n--- Procesando lote: archivos {inicio} al {fin} ---")
        resultados_lote = []
        
        for index, row in tqdm(lote_df.iterrows(), total=len(lote_df)):
            fold_dir = f"fold{row['fold']}"
            nombre_archivo = row['slice_file_name']
            
            # Construir la ruta correcta del audio
            ruta_audio = os.path.join(ruta_base, "UrbanSound8K", "audio", fold_dir, nombre_archivo)
            
            if not os.path.exists(ruta_audio):
                continue
                
            try:
                # 1. Cargar señal de audio
                y, sr = librosa.load(ruta_audio, sr=None)
                
                # 2. Cálculo de la STFT
                stft_matrix = librosa.stft(y, n_fft=2048, hop_length=512)
                stft_magnitude = np.abs(stft_matrix)
                
                # 3. Extraer estadísticos de la matriz STFT para resumirla en valores tabulares
                resultados_lote.append({
                    'slice_file_name': nombre_archivo,
                    'fold': row['fold'],
                    'classID': row['classID'],
                    'class': row['class'],
                    'stft_mean': np.mean(stft_magnitude),
                    'stft_std': np.std(stft_magnitude),
                    'stft_max': np.max(stft_magnitude),
                    'stft_min': np.min(stft_magnitude)
                })
            except Exception as e:
                print(f"Error procesando STFT para {nombre_archivo}: {e}")
        
        # Convertir el lote actual a DataFrame
        df_lote = pd.DataFrame(resultados_lote)
        
        # Guardar en el CSV de forma incremental (añade al archivo sin sobrescribir)
        if not archivo_csv_existe:
            df_lote.to_csv(ruta_salida_csv, index=False, mode='w')
            archivo_csv_existe = True
        else:
            df_lote.to_csv(ruta_salida_csv, index=False, mode='a', header=False)
            
        print(f"Lote guardado exitosamente en: {ruta_salida_csv}")

    tiempo_fin = time.perf_counter()
    tiempo_total = tiempo_fin - tiempo_inicio
    
    print(f"\n¡Proceso completado con éxito!")
    print(f"Tiempo Total (T1 STFT): {tiempo_total:.4f} segundos ({tiempo_total/60:.2f} minutos)")

# Rutas de trabajo configuradas en tu laptop
ruta_base = r"C:\Users\eliye\OneDrive\Documentos\SextoSemestre\ComputoParalelo\Proyecto-Paralelo"
ruta_salida_csv = os.path.join(ruta_base, "resultados_stft_baseline.csv")

# Ejecutar la función
extraer_stft_y_guardar_csv(ruta_base, df, ruta_salida_csv, tamaño_lote=500)
```

```python
import os
import time
import numpy as np
import pandas as pd
import librosa
from tqdm.auto import tqdm

def extraer_mfcc_y_guardar_csv(ruta_base, df_metadata, ruta_salida_csv, tamaño_lote=500):
   
    "Extrae características de los Coeficientes Cepstrales en Frecuencias Mel (MFCC) de forma secuencial"
    tiempo_inicio = time.perf_counter()
    archivo_csv_existe = False
    
    total_archivos = len(df_metadata)
    
    # Procesar por lotes para proteger la memoria RAM
    for inicio in range(0, total_archivos, tamaño_lote):
        fin = min(inicio + tamaño_lote, total_archivos)
        lote_df = df_metadata.iloc[inicio:fin]
        
        print(f"\n--- Procesando lote MFCC: archivos {inicio} al {fin} ---")
        resultados_lote = []
        
        for index, row in tqdm(lote_df.iterrows(), total=len(lote_df)):
            fold_dir = f"fold{row['fold']}"
            nombre_archivo = row['slice_file_name']
            
            # Construir la ruta correcta del audio
            ruta_audio = os.path.join(ruta_base, "UrbanSound8K", "audio", fold_dir, nombre_archivo)
            
            if not os.path.exists(ruta_audio):
                continue
                
            try:
                # 1. Cargar señal de audio
                y, sr = librosa.load(ruta_audio, sr=None)
                
                # 2. Cálculo de los MFCC (por defecto se extraen 13 coeficientes)
                mfcc_matrix = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13, n_fft=2048, hop_length=512)
                
                # 3. Extraer estadísticos de la matriz MFCC para resumirla en valores tabulares
                resultados_lote.append({
                    'slice_file_name': nombre_archivo,
                    'fold': row['fold'],
                    'classID': row['classID'],
                    'class': row['class'],
                    'mfcc_mean': np.mean(mfcc_matrix),
                    'mfcc_std': np.std(mfcc_matrix),
                    'mfcc_max': np.max(mfcc_matrix),
                    'mfcc_min': np.min(mfcc_matrix)
                })
            except Exception as e:
                print(f"Error procesando MFCC para {nombre_archivo}: {e}")
        
        # Convertir el lote actual a DataFrame
        df_lote = pd.DataFrame(resultados_lote)
        
        # Guardar en el CSV de forma incremental (añade al archivo sin sobrescribir)
        if not archivo_csv_existe:
            df_lote.to_csv(ruta_salida_csv, index=False, mode='w')
            archivo_csv_existe = True
        else:
            df_lote.to_csv(ruta_salida_csv, index=False, mode='a', header=False)
            
        print(f"Lote guardado exitosamente en: {ruta_salida_csv}")

    tiempo_fin = time.perf_counter()
    tiempo_total = tiempo_fin - tiempo_inicio
    
    print(f"\n¡Extracción de MFCC completada con éxito!")
    print(f"Tiempo Total (T1 MFCC): {tiempo_total:.4f} segundos ({tiempo_total/60:.2f} minutos)")

# Rutas de trabajo configuradas en tu laptop
ruta_base = r"C:\Users\eliye\OneDrive\Documentos\SextoSemestre\ComputoParalelo\Proyecto-Paralelo"
ruta_salida_csv = os.path.join(ruta_base, "resultados_mfcc_baseline.csv")

# Ejecutar la función
extraer_mfcc_y_guardar_csv(ruta_base, df, ruta_salida_csv, tamaño_lote=500)
```

```python
%%writefile pipeline_stft_hibrido.py
import os
import time
import numpy as np
import pandas as pd
import librosa
from mpi4py.MPI import COMM_WORLD
from multiprocessing import Pool
import socket

def extraer_stft_individual(row_data):
    ruta_base, fold, nombre_archivo, class_id, clase_nombre = row_data
    ruta_audio = os.path.join(ruta_base, "UrbanSound8K", "audio", f"fold{fold}", nombre_archivo)
    
    if not os.path.exists(ruta_audio):
        return None
        
    try:
        y, sr = librosa.load(ruta_audio, sr=None)
        stft_matrix = librosa.stft(y, n_fft=2048, hop_length=512)
        stft_magnitude = np.abs(stft_matrix)
        
        return {
            'slice_file_name': nombre_archivo,
            'fold': fold,
            'classID': class_id,
            'class': clase_nombre,
            'stft_mean': np.mean(stft_magnitude),
            'stft_std': np.std(stft_magnitude),
            'stft_max': np.max(stft_magnitude),
            'stft_min': np.min(stft_magnitude)
        }
    except Exception as e:
        return None

def main():
    comm = COMM_WORLD
    rank = comm.Get_rank()
    size = comm.Get_size()
    hostname = socket.gethostname()

    comm.Barrier()
    tiempo_inicio = comm.Wtime()

    ruta_base = r"C:\Users\eliye\OneDrive\Documentos\SextoSemestre\ComputoParalelo\Proyecto-Paralelo"
    ruta_csv = os.path.join(ruta_base, "UrbanSound8K", "metadata", "UrbanSound8K.csv")

    if rank == 0:
        print(f"Procesos MPI activos: {size} | Máquina: {hostname}")
        df_meta = pd.read_csv(ruta_csv)
        
        # Empaquetamos los datos necesarios de cada fila
        registros = list(zip(
            [ruta_base]*len(df_meta), 
            df_meta['fold'], 
            df_meta['slice_file_name'],
            df_meta['classID'],
            df_meta['class']
        ))
    else:
        registros = None

    registros = comm.bcast(registros, root=0)

    #  Reparto MPI
    lotes_por_proceso = np.array_split(registros, size)
    mi_lote = lotes_por_proceso[rank].tolist()

    print(f"-> [MPI Rank {rank}] Procesando {len(mi_lote)} archivos (STFT).")

    # Reparto Multiprocessing
    num_hilos_internos = 2
    resultados_locales = []

    with Pool(processes=num_hilos_internos) as pool:
        resultados_locales = pool.map(extraer_stft_individual, mi_lote)

    resultados_locales = [res for res in resultados_locales if res is not None]

    comm.Barrier()
    tiempo_fin = comm.Wtime()
    tiempo_local = tiempo_fin - tiempo_inicio

    todos_los_tiempos = comm.gather(tiempo_local, root=0)
    todos_los_resultados = comm.gather(resultados_locales, root=0)

    if rank == 0:
        print(f"\n[Resultados de Rendimiento - STFT]:")
        for p_id, t in enumerate(todos_los_tiempos):
            print(f"  • Proceso MPI {p_id}: {t:.4f} segundos")
        
        tiempo_maximo = max(todos_los_tiempos)
        print(f"\n> Tiempo Total Paralelo (STFT): {tiempo_maximo:.4f} segundos ({tiempo_maximo/60:.2f} minutos)")
        
        resultados_totales = [item for sublist in todos_los_resultados for item in sublist]
        df_final = pd.DataFrame(resultados_totales)
        ruta_salida = os.path.join(ruta_base, "resultados_stft_hibrido.csv")
        df_final.to_csv(ruta_salida, index=False)
        print(f"¡Guardado exitosamente en: {ruta_salida}!")

if __name__ == '__main__':
    main()
```
